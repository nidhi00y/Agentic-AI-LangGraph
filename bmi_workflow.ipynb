```python
from langgraph.graph import StateGraph,START,END 
from typing import TypedDict
```

```python
class BMISTATE(TypedDict):
    weight_kg:float
    height_m:float
    bmi:float
    bmi_label:str
```

```python
def calculate_bmi(state:BMIState) -> BMIState:
    weight = state['weight_kg']
    height = state['height_m']

    bmi = weight/(height**2)

    state['bmi'] = bmi
```

```python
def bmi_label(state:BMIState) -> BMIState:
    bmi = state['bmi']
    if bmi < 18.5:
        label = 'Underweight'
    elif bmi < 25:
        label = 'Normal weight'
    elif bmi < 30:
        label = 'Overweight'
    else:
        label = 'Obesity'

    state['bmi_label'] = label
```

```python
#define graph
graph = StateGraph(BMIState)

#add nodes
graph.add_node('calculate_bmi',calculate_bmi)
graph.add_node('bmi_label',bmi_label)

#add edges
graph.add_edge(START,'calculate_bmi')
graph.add_edge('calculate_bmi','bmi_label')
graph.add_edge('bmi_label',END)

#compile graph
workflow = graph.compile()
```

```python
#execute graph
output_state = workflow.invoke({'weight_kg':53,'height_m':1.7})
print(output_state)
```
